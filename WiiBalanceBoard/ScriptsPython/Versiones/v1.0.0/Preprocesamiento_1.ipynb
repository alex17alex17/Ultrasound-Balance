```python
import pandas as pd
import numpy as np
from scipy.interpolate import interp1d

# 🧠 Simulación: datos desigualmente espaciados (como los que mediría la Wii Balance Board)
tiempo = np.array([0.00, 0.011, 0.021, 0.033, 0.045, 0.058, 0.071])
copx = np.array([1.0, 1.2, 1.5, 1.3, 1.4, 1.6, 1.7])
copy = np.array([3.2, 3.5, 3.6, 3.7, 3.8, 4.0, 4.2])

# 🎯 Definir el nuevo eje temporal (100 Hz = 0.01 s)
t_uniforme = np.arange(0, tiempo[-1], 0.01)

# 🔹 Crear función de interpolación lineal
interp_copx = interp1d(tiempo, copx, kind='linear', fill_value="extrapolate")
interp_copy = interp1d(tiempo, copy, kind='linear', fill_value="extrapolate")

# 📈 Calcular los valores interpolados
copx_uniforme = interp_copx(t_uniforme)
copy_uniforme = interp_copy(t_uniforme)

# 📊 Guardar en un DataFrame
df_interp = pd.DataFrame({
    "t": t_uniforme,
    "COPx": copx_uniforme,
    "COPy": copy_uniforme
})

print(df_interp.head(10))
```
